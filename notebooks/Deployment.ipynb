# Deployment simulado – chamar todos os modelos guardados em `models/`

Fase de Deployment (CRISP-ML). Este notebook usa os módulos de `src/` e é a versão em notebook do `main.py`.

1. Carregar o modelo campeão (via `models_manifest.json`).
2. Carregar **todos** os modelos guardados (7 de scikit-learn + 1 MLP TensorFlow).
3. Classificar uma fraude real e uma transação legítima real do conjunto de teste.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))  # raiz do projeto, para importar src/

import pandas as pd
from src.config import MODELS_DIR, TEST_FILE
from src.predict import PredictorFraude, PredictorTodos
```

## 1. Modelo campeão

```python
predictor = PredictorFraude(MODELS_DIR)
print('Campeão:', predictor.nome_campeao, '->', predictor.ficheiro_campeao)
```

## 2. Todos os modelos guardados

```python
todos = PredictorTodos(MODELS_DIR)
print(len(todos.modelos), 'modelos carregados:', list(todos.modelos))
```

## 3. Exemplos reais do conjunto de teste

```python
teste = pd.read_csv(TEST_FILE)
fraude = teste[teste['Class'] == 1].iloc[0].drop('Class').to_dict()
legitima = teste[teste['Class'] == 0].iloc[0].drop('Class').to_dict()
```

```python
print('Transação FRAUDULENTA (real) - campeão:', predictor.prever(fraude))
todos.prever(fraude)
```

```python
print('Transação LEGÍTIMA (real) - campeão:', predictor.prever(legitima))
todos.prever(legitima)
```
